```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from statsmodels.tsa.stattools import adfuller
```

```python
df = pd.read_excel("Draft Data ADW.xlsx")
df.head()
```

```python
plt.figure(figsize=(14, 6))
plt.plot(df['Minggu Ke-'], df['Total Billing'] / 1e6, label='Total Billing Mingguan', color='#1f77b4', alpha=0.6, linewidth=1)

plt.title('Time Series Plot Total Billing Mingguan Rumah Sakit Tahun 2025', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Tanggal', fontsize=12)
plt.ylabel('Total Billing', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(loc='upper right', fontsize=11)
plt.tight_layout()
```

```python
def adf_test(df):
    hasil = adfuller(df["Total Billing"])

    print("ADF Statistic :", hasil[0])
    print("p-value       :", hasil[1])
    print("Critical Value:")

    for key, value in hasil[4].items():
        print(f"   {key} : {value}")

    if hasil[1] < 0.05:
        print("\nKesimpulan: Data STASIONER")
    else:
        print("\nKesimpulan: Data BELUM STASIONER")

adf_test(df)
```
